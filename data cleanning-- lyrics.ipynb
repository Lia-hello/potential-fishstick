```python
# Billboard Lyrics Dataset Cleaning Code
# For Research: Lyric Diversity Before and After 2023
# Data Source: Billboard Top Charts

import pandas as pd
import re
import os

# --------------------------
# Load raw Billboard data
# --------------------------
df = pd.read_csv("billboard_24years_lyrics_spotify.csv")

# --------------------------
# Step 1: Keep only necessary columns
# --------------------------
# Adjust column names based on your actual CSV columns
df = df[["song", "band_singer", "titletext", "year", "lyrics"]]
df.columns = ["song", "band_singer", "titletext", "year", "lyrics"]

# --------------------------
# Step 2: Remove missing values
# --------------------------
df = df.dropna(subset=["song", "band_singer", "year", "lyrics", "titletext"])

# --------------------------
# Step 3: Filter year range (2015–2024)
# --------------------------
df["year"] = df["year"].astype(int)
df = df[(df["year"] >= 2015) & (df["year"] <= 2024)]



# --------------------------
# Step 5: Clean lyrics text
# --------------------------
def clean_lyrics(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\n', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df["cleaned_lyrics"] = df["lyrics"].apply(clean_lyrics)

```

```python
# --------------------------
# Step 4: Clean lyrics text
# --------------------------
def clean_lyrics(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\n', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df["cleaned_lyrics"] = df["lyrics"].apply(clean_lyrics)

```

```python
# --------------------------
# Step 5: Remove duplicates
# --------------------------
df = df.drop_duplicates(subset=["song", "band_singer"], keep="first")
```

```python
# --------------------------
# Step 6: Add AI period label (2023 split)
# --------------------------
df["ai_period"] = df["year"].apply(lambda x: "Post-2023" if x >= 2023 else "Pre-2023")
```

```python
# --------------------------
# Step 7: Save cleaned dataset
# --------------------------
os.makedirs("../data", exist_ok=True)
df.to_csv("../data/cleaned_billboard_dataset.csv", index=False)
```
