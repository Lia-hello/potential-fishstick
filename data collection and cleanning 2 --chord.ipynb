```python
import pandas as pd

# Load dataset
df = pd.read_csv("data/chordonomicon_v2.csv")

# Check column names first
print(df.columns)
```

```python
 # =========================
# 1. Extract year
# =========================

# Assuming the date column is called 'release_date'
df['year'] = pd.to_datetime(
    df['release_date'],
    errors='coerce'
).dt.year
```

```python
# =========================
# 2. Keep only needed columns
# =========================

clean_df = df[[
    'year',      # release year
    'song',      # song title
    'chords'     # chord progression
]]
```

```python
# =========================
# 3. Remove missing values
# =========================

clean_df = clean_df.dropna(
    subset=['year', 'song', 'chords']
)
```

```python
# =========================
# 4. Convert year to integer
# =========================

clean_df['year'] = clean_df['year'].astype(int)
```

```python
# =========================
# 5. Remove empty strings
# =========================

clean_df = clean_df[
    (clean_df['song'].str.strip() != '') &
    (clean_df['chords'].str.strip() != '')
]
```

```python
# =========================
# 6. Remove duplicates
# =========================

clean_df = clean_df.drop_duplicates()
```

```python
# =========================
# 7. Preview cleaned data
# =========================

print(clean_df.head())

print("\nDataset shape:")
print(clean_df.shape)
```

```python
# =========================
# 8. Save cleaned dataset
# =========================

clean_df.to_csv(
    "clean_chord_data.csv",
    index=False
)

print("\nData cleaning completed.")
```
